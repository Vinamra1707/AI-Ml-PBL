```python
import pandas as pd
```

```python
df=pd.read_csv("cleaned_dataset.csv")
display(df)
```

```python
from sklearn.preprocessing import StandardScaler

X = df.drop(columns=['Timestamp', 'Date', 'Energy_Usage_kWh'])
y = df['Energy_Usage_kWh']

print("                 FEATURES BEFORE SCALING                 ")
print(X.head(3))
```

```python
print("                     RANGES BEFORE SCALING                       ")
for col in X.columns:
    print(f"  {col}: min={round(X[col].min(),2)}, max={round(X[col].max(),2)}")
```

```python
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

X_scaled = pd.DataFrame(X_scaled, columns=X.columns)

print("                     RANGES AFTER SCALING                        ")
for col in X_scaled.columns:
    print(f"  {col}: min={round(X_scaled[col].min(),2)}, max={round(X_scaled[col].max(),2)}")

print()
```
